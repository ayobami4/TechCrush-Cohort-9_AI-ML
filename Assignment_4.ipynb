**Flexible Greeting Function**

Write a function called `greet_person()` that takes a name as a parameter and prints "Hello, name! Welcome to Python!"

- Allow multiple names using *args.

- If no name is given, print "Hello, Guest! Welcome to Python!"
If a title (e.g., "Dr.", "Prof.", "King") is provided using **kwargs, include it in the greeting.

```python
# Defining the greeting function
def greet_person(*args, **kwargs):

    # Checking whether any name was provided
    if not args:
        print('Hello, Guest! Welcome to Python!')
        return

    # Getting the first value supplied through a keyword argument
    title = next(iter(kwargs.values()), '')

    # Looping through each name provided
    for name in args:

        # Checking whether a title was provided
        if title:
            print(f'Hello, {title} {name}! Welcome to Python!')

        # Greeting the person by name only, if no title was provided
        else:
            print(f'Hello, {name}! Welcome to Python!')
```

```python
# Testing the function with one name and no title
greet_person('Orebanjo')

# Testing the function with multiple names
greet_person('Ayomide', 'Solomon')

# Testing the function with a title
greet_person('Orebanjo', title='Dr.')

# Testing the function with no name
greet_person()
```
