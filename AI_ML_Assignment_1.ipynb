### Part 3: Basic Python Operations
- Given two numbers a = 30 and b = 8, perform the following operations and print the results:

  - Addition
  - Floor division
  - Exponentiation

- Write a Python program that swaps the values of two variables without using a third variable.

```python
# Initializing the variables
a = 30
b = 8

# Performing addition
addition = a + b

# Performing floor division
floor_division = a // b

# Performing exponentiation
exponentiation = a ** b

# Printing the results
print(f'The sum of {a} and {b} is {addition}')
print(f'The floor division of {a} and {b} is {floor_division}')
print(f'The exponentiation of {a} and {b} is {exponentiation}')
```

### Part 4: Logical & Comparison Operators

- Write a Python program that checks whether the defined variables are greater than 10 and even.

```python
x = 30
y = 25
z = 10

# Checking whether the defined variables are greater than 10 and even
if x > 10 and x % 2 == 0:
  print(f'{x} is greater than 10 and even')
else:
  print(f'{x} is not greater than 10 and even')

if y > 10 and y % 2 == 0:
  print(f'{y} is greater than 10 and even')
else:
  print(f'{y} is not greater than 10 and even')

if z > 10 and z % 2 == 0:
  print(f'{z} is greater than 10 and even')
else:
  print(f'{z} is not greater than 10 and even')
```

- Hence, Write conditions using logical operators to:

  - Check if x is greater than y and z.
  - Check if at least one of the variables is less than 5.

```python
# Checking if x is greater than y and z
if x > y and x > z:
  print(f'{x} is greater than {y} and {z}')
else:
  print(f'{x} is not greater than {y} and {z}')

# Checking if at least one of the variables is less than 5
if x < 5 or y < 5 or z < 5:
  print('At least one of the variables is less than 5')
else:
  print('None of the variables is less than 5')
```
