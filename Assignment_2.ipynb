**Hands-on Practice**

1. Create a 3 x 4 NumPy array with random integers between 1 and 100. Display the array.
2. From the array created in Question 1, extract the following: (a) the first two rows. (b) the last 3 columns (c) All even numbers in the array.
3. Perform the following transformation on the array created in Question 1: (a) reshape the array into a 3D (2, 2, 3) array. (b) Now you flatten the reshaped array back to 1D.
4. Generate a 4x4 identity matrix and replace the diagonal with numbers from 10 to 40, with a step of 10.
5. Perform matrix multiplication between a 3x4 matrix of random integers (1-10) and a 4x3 matrix of random integers (1-10).
6. Create a function that takes a 2D NumPy array as input and then returns a new array where all even numbers are squared and all odds numbers are replaced by -1

```python
# Creating a 3 x 4 NumPy array with random integers between 1 and 100
my_array = np.random.randint(1, 101, (3, 4))

# Displaying the array
print(my_array)
```

```python
# Extracting the first two rows
first_two_rows = my_array[:2, :]
print(f'The first two rows: \n {first_two_rows} \n')

# Extracting the last 3 columns
last_three_columns = my_array[:, -3:]
print(f'The last 3 columns: \n {last_three_columns} \n')

# Extracting all even numbers in the array
even_numbers = my_array[my_array % 2 == 0]
print(f'All the even numbers: \n {even_numbers}')
```

```python
# Reshaping the array into a 3D (2, 2, 3) array
reshaped_array = my_array.reshape(2, 2, 3)
print(f'Reshaped array: \n {reshaped_array} \n')

# Flattening the reshaped array back to 1D
flattened_array = reshaped_array.flatten()
print(f'Flattened array: \n {flattened_array}')
```

```python
# Generating a 4x4 identity matrix
identity_matrix = np.eye(4)
print(f'Identity matrix: \n {identity_matrix} \n')

# Replacing the diagonal with numbers from 10 to 40, with a step of 10
diagonal_matrix = np.diag(np.arange(10, 41, 10))
print(f'Replaced diagonal: \n {diagonal_matrix}')
```

```python
# Performing matrix multiplication between a 3x4 matrix of random integers (1-10) and a 4x3 matrix of random integers (1-10)
# Defining the matrix variables
matrix1 = np.random.randint(1, 11, (3, 4))
matrix2 = np.random.randint(1, 11, (4, 3))

# Performing multiplication between the two defined matrix variables
multiply = np.dot(matrix1, matrix2)

# Printing the result
print(f'Result: {multiply}')
```

```python
# Creating a function that takes a 2D NumPy array as input and then returns a new array where all even numbers are squared and all odds numbers are replaced by -1
def my_function(my_2darray):
  result = np.where(my_2darray % 2 == 0, my_2darray ** 2, -1)
  return result

# Defining a 2D NumPy array
my_array = np.array([[1, 2, 3], [4, 5, 6]])

# Calling the function
my_function(my_array)
```
