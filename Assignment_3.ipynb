**Task 1: Implement a Custom Calculator**

Write a function calculate() that takes three parameters:

-  num1: The first number.
-  num2: The second number.
-  operation: A string specifying the operation ("add", "subtract", "multiply", "divide").

The function should:

- Perform the specified mathematical operation.
- Return the result.
- Handle division by zero gracefully.

**Note: Allow users to pass numbers as positional or keyword arguments.**

**Solution**

```python
# Creating a function that takes three parameters and performs simple arithmetic operations
def calculate(num1, num2, operation):
    if operation == 'add':
        return num1 + num2
    elif operation == 'subtract':
        return num1 - num2
    elif operation == 'multiply':
        return num1 * num2
    elif operation == 'divide':
      if num2 != 0:
        return num1 / num2
      else:
        return 'Error: Division by zero'
```

```python
# Printing and calling the function
print('Addition:', calculate(10, 2, 'add'))
print('Subtraction:', calculate(10, 2, 'subtract'))
print('Multiplication:', calculate(10, 2, 'multiply'))
print('Division:', calculate(10, 2, 'divide'))
```
